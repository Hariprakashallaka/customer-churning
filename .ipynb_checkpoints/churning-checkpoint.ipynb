```python
def evaluate_tree(max_depth=None,
                  min_samples_split=2,
                  min_samples_leaf=1):

    model = DecisionTreeClassifier(
        criterion="gini",
        max_depth=max_depth,
        min_samples_split=min_samples_split,
        min_samples_leaf=min_samples_leaf,
        random_state=42
    )

    model.fit(X_train_processed, y_train)

    train_pred = model.predict(X_train_processed)
    test_pred = model.predict(X_test_processed)

    train_acc = accuracy_score(y_train, train_pred)
    test_acc = accuracy_score(y_test, test_pred)

    return (
        train_acc,
        test_acc,
        model.get_depth(),
        model.get_n_leaves()
    )
```

```python
split_values = [2, 5, 10, 20, 50, 100]

for value in split_values:

    result = evaluate_tree(
        max_depth=None,
        min_samples_split=value,
        min_samples_leaf=1
    )

    print(
        f"min_samples_split={value:3d} | "
        f"Train={result[0]:.4f} | "
        f"Test={result[1]:.4f} | "
        f"Depth={result[2]} | "
        f"Leaves={result[3]}"
    )
```

```python
leaf_values = [1, 2, 5, 10, 20, 50]

for value in leaf_values:

    result = evaluate_tree(
        max_depth=None,
        min_samples_split=2,
        min_samples_leaf=value
    )

    print(
        f"min_samples_leaf={value:3d} | "
        f"Train={result[0]:.4f} | "
        f"Test={result[1]:.4f} | "
        f"Depth={result[2]} | "
        f"Leaves={result[3]}"
    )
```

```python
path = dt_model.cost_complexity_pruning_path(
    X_train_processed,
    y_train
)
ccp_alphas = path.ccp_alphas

print(ccp_alphas)
```

```python
for alpha in ccp_alphas:

    model = DecisionTreeClassifier(
        criterion="gini",
        ccp_alpha=alpha,
        random_state=42
    )

    model.fit(X_train_processed, y_train)

    train_pred = model.predict(X_train_processed)
    test_pred = model.predict(X_test_processed)

    print(
        f"alpha={alpha:.6f} | "
        f"Train={accuracy_score(y_train, train_pred):.4f} | "
        f"Test={accuracy_score(y_test, test_pred):.4f} | "
        f"Depth={model.get_depth()} | "
        f"Leaves={model.get_n_leaves()}"
    )
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
import pandas as pd


# 1. Create Decision Tree
dt = DecisionTreeClassifier(
    random_state=42
)


# 2. Hyperparameter values
param_grid = {
    "criterion": ["gini", "entropy"],
    "max_depth": [5, 7, 9, 11, 13],
    "min_samples_split": [2, 10, 20],
    "min_samples_leaf": [1, 2, 5]
}


# 3. GridSearchCV
grid_search = GridSearchCV(
    estimator=dt,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)


# 4. Train GridSearchCV
grid_search.fit(
    X_train_processed,
    y_train
)


# 5. Best parameters
print("Best Parameters:")
print(grid_search.best_params_)


# 6. Best CV score
print("\nBest Cross-Validation Accuracy:")
print(grid_search.best_score_)


# 7. Best model
best_dt = grid_search.best_estimator_

print("\nBest Decision Tree:")
print(best_dt)


# 8. Predict test data
y_pred_best = best_dt.predict(
    X_test_processed
)


# 9. Final test accuracy
print("\nFinal Test Accuracy:")
print(accuracy_score(y_test, y_pred_best))


# 10. Tree complexity
print("\nTree Depth:")
print(best_dt.get_depth())

print("\nNumber of Leaves:")
print(best_dt.get_n_leaves())


# 11. Classification report
print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred_best,
        target_names=["No Churn", "Churn"]
    )
)


# 12. Confusion Matrix
print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_test,
        y_pred_best
    )
)


# 13. Feature Importance
feature_importance = pd.Series(
    best_dt.feature_importances_,
    index=feature_names
).sort_values(ascending=False)

print("\nFeature Importance:")
print(feature_importance)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# --------------------------------------------------
# 1. Base Random Forest
# --------------------------------------------------

rf = RandomForestClassifier(
    random_state=42,
    n_jobs=-1
)


# --------------------------------------------------
# 2. Parameter Grid
# --------------------------------------------------

param_grid = {
    "n_estimators": [50, 100],
    "max_depth": [None, 10, 15],
    "min_samples_split": [2, 5],
    "min_samples_leaf": [1, 2],
    "max_features": [None, "sqrt"]
}


# --------------------------------------------------
# 3. GridSearchCV
# --------------------------------------------------

grid_search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1,
    verbose=2
)


# --------------------------------------------------
# 4. Train
# --------------------------------------------------

grid_search.fit(
    X_train_processed,
    y_train
)


# --------------------------------------------------
# 5. Best Parameters
# --------------------------------------------------

print("Best Parameters:")
print(grid_search.best_params_)


# --------------------------------------------------
# 6. Best CV Score
# --------------------------------------------------

print(
    "\nBest Cross-Validation Accuracy:",
    grid_search.best_score_
)


# --------------------------------------------------
# 7. Best Model
# --------------------------------------------------

best_rf = grid_search.best_estimator_


# --------------------------------------------------
# 8. Test Prediction
# --------------------------------------------------

test_pred = best_rf.predict(
    X_test_processed
)


# --------------------------------------------------
# 9. Final Test Accuracy
# --------------------------------------------------

final_accuracy = accuracy_score(
    y_test,
    test_pred
)

print(
    "\nFinal Test Accuracy:",
    final_accuracy
)


# --------------------------------------------------
# 10. Classification Report
# --------------------------------------------------

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        test_pred
    )
)


# --------------------------------------------------
# 11. Confusion Matrix
# --------------------------------------------------

cm = confusion_matrix(
    y_test,
    test_pred
)

print("\nConfusion Matrix:")
print(cm)


# --------------------------------------------------
# 12. Confusion Matrix Visualization
# --------------------------------------------------

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=["No Churn", "Churn"],
    yticklabels=["No Churn", "Churn"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Final Random Forest Confusion Matrix")

plt.show()


# --------------------------------------------------
# 13. Top GridSearch Results
# --------------------------------------------------

results = pd.DataFrame(
    grid_search.cv_results_
)

top_results = results[
    [
        "params",
        "mean_test_score",
        "std_test_score",
        "rank_test_score"
    ]
].sort_values(
    "rank_test_score"
).head(10)

print("\nTop 10 Parameter Combinations:")
print(top_results)
```

```python
# 1. Remove Payment Delay
X_no_payment_delay = df.drop(
    columns=["Churn", "Payment Delay"]
)

y = df["Churn"]


# 2. Train/Test Split
from sklearn.model_selection import train_test_split

X_train_no_pd, X_test_no_pd, y_train_no_pd, y_test_no_pd = train_test_split(
    X_no_payment_delay,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)


# 3. Identify columns
numerical_features_no_pd = X_train_no_pd.select_dtypes(
    include=["int64", "float64"]
).columns

categorical_features_no_pd = X_train_no_pd.select_dtypes(
    include=["object"]
).columns


# 4. Preprocessing
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

preprocessor_no_pd = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numerical_features_no_pd
        ),
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features_no_pd
        )
    ]
)


X_train_no_pd_processed = preprocessor_no_pd.fit_transform(
    X_train_no_pd
)

X_test_no_pd_processed = preprocessor_no_pd.transform(
    X_test_no_pd
)


# 5. Random Forest
from sklearn.ensemble import RandomForestClassifier

rf_no_pd = RandomForestClassifier(
    n_estimators=100,
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=2,
    max_features=None,
    random_state=42,
    n_jobs=-1
)

rf_no_pd.fit(
    X_train_no_pd_processed,
    y_train_no_pd
)


# 6. Predictions
train_pred_no_pd = rf_no_pd.predict(
    X_train_no_pd_processed
)

test_pred_no_pd = rf_no_pd.predict(
    X_test_no_pd_processed
)


# 7. Accuracy
from sklearn.metrics import accuracy_score

print(
    "Training Accuracy:",
    accuracy_score(y_train_no_pd, train_pred_no_pd)
)

print(
    "Testing Accuracy :",
    accuracy_score(y_test_no_pd, test_pred_no_pd)
)


# 8. Classification Report
from sklearn.metrics import classification_report

print("\nClassification Report:")
print(
    classification_report(
        y_test_no_pd,
        test_pred_no_pd
    )
)


# 9. Confusion Matrix
from sklearn.metrics import confusion_matrix

print("Confusion Matrix:")
print(
    confusion_matrix(
        y_test_no_pd,
        test_pred_no_pd
    )
)
```
